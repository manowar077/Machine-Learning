When should you use MinMaxScaler?

It's especially useful when your ML algorithm is sensitive to the scale of features.

Good cases:

KNN ⭐⭐⭐⭐⭐

K-Means clustering ⭐⭐⭐⭐⭐

Neural Networks ⭐⭐⭐⭐⭐

SVM ⭐⭐⭐⭐

Gradient Descent-based models ⭐⭐⭐⭐

PCA can also benefit from scaling.

For example, in KNN:

Distance = √((Age difference)² + (Salary difference)²)

Because salary has much larger numbers, it can dominate the distance.

After normalization, both features have comparable scales.

2. Example with KNN

Suppose we want to find which student is closest to a new student.

New student:

Age = 24
Salary = 45,000

Student A:

Age = 20
Salary = 20,000

Euclidean distance:

$$ Distance = \sqrt{(24-20)^2 + (45000-20000)^2} $$

So:

$$ = \sqrt{4^2 + 25000^2} $$

The salary difference is 25,000, while the age difference is only 4.

Therefore, salary dominates the distance.

This can be a problem because the algorithm doesn't know that salary is "more important" just because its numerical values are larger.
