**Write a program that will reverse a four digit number also check whether the reverse is true**

```python
num = input()

if len(num) == 4 and num.isdigit():
    
    reversed_num = num[::-1]
    print(reversed_num)

    if reversed_num == num:
        print("Palindrome")
    else:
        print("Not a palindrome")

else:
    print("Please enter a valid 4-digit number")
```
