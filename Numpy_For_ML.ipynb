**Broadcasting**

**Broadcasting** in NumPy is the mechanism that lets NumPy perform arithmetic operations on arrays of **different shapes** without explicitly copying data — it automatically "stretches" the smaller array across the larger one so their shapes become compatible.

**The problem it solves**

Without broadcasting, operations like `array + scalar` or adding arrays of different shapes would require you to manually reshape/tile the smaller array to match. Broadcasting removes that boilerplate.

```python
import numpy as np

a = np.array([1, 2, 3])
b = 5
print(a + b)  # [6 7 8]
```

Here, `5` is conceptually "broadcast" to `[5, 5, 5]` and added elementwise — but NumPy doesn't actually create that array in memory; it's done efficiently under the hood.

**The broadcasting rules**

When operating on two arrays, NumPy compares their shapes **element-wise, starting from the trailing (rightmost) dimension**. Two dimensions are compatible when:
1. They are **equal**, OR
2. One of them is **1** (or missing entirely)

If neither condition holds for any dimension, you get a `ValueError: operands could not be broadcast together`.

**Example — vector + matrix**

```python
A = np.array([[1, 2, 3],
              [4, 5, 6]])      # shape (2, 3)

b = np.array([10, 20, 30])     # shape (3,)

print(A + b)
# [[11 22 33]
#  [14 25 36]]
```

Here, `b` (shape `(3,)`) is broadcast across each row of `A` (shape `(2,3)`), because trailing dimensions match (`3 == 3`), and the missing leading dimension in `b` is treated as `1`, which stretches to match `2`.

**Example — column vector + row vector**

```python
col = np.array([[1], [2], [3]])   # shape (3, 1)
row = np.array([10, 20, 30])      # shape (3,)

print(col + row)
# [[11 21 31]
#  [12 22 32]
#  [13 23 33]]
```

This produces a `(3,3)` result — both arrays get stretched along their size-1 dimension to form an outer-product-like grid. This pattern is genuinely useful (e.g., building distance matrices, outer sums).

**Visualizing the shape-matching**

| Array | Shape |
|---|---|
| A | `(2, 3)` |
| B | `(3,)` → treated as `(1, 3)` |
| Result | `(2, 3)` |

```
A:      2 x 3
B:          3   (implicitly 1 x 3)
------------
Result: 2 x 3
```

**Why it matters practically**
- **Performance:** avoids creating unnecessary copies of data — broadcasting is implemented in C and doesn't allocate the "stretched" array in memory.
- **Cleaner code:** you write `A + b` instead of manually tiling `b` with `np.tile()`.
- Extremely common in data science / ML code: normalizing data (`(X - mean) / std`), adding bias vectors in neural nets, computing pairwise distances, etc.

**Common failure case**

```python
a = np.zeros((3, 4))
b = np.zeros((3,))
a + b  # ValueError!
```

This fails because trailing dimensions are `4` vs `3` — neither equal nor `1`. You'd need `b` reshaped to `(3, 1)` (via `b.reshape(3,1)` or `b[:, None]`) if you wanted it to broadcast against the rows instead of columns.

Want me to show a practical example — like normalizing a dataset or computing a pairwise distance matrix — where broadcasting actually saves real code and runtime?

# Vectorization in NumPy

**Vectorization** means performing operations on an entire array at once instead of using Python loops.

## Without Vectorization ❌

```python
prices = [100, 200, 300]
discount = 10

final_prices = []

for price in prices:
    final_prices.append(price - (price * discount / 100))

print(final_prices)
```

Here, Python processes each element individually:

```text
100 → calculate
200 → calculate
300 → calculate
```

Output:

```text
[90.0, 180.0, 270.0]
```

---

## With Vectorization ✅

```python
import numpy as np

prices = np.array([100, 200, 300])
discount = 10

final_prices = prices - (prices * discount / 100)

print(final_prices)
```

Output:

```text
[ 90. 180. 270.]
```

Here, you write one operation for the entire array.

```text
[100, 200, 300]
        ↓
   One operation
        ↓
[90, 180, 270]
```

## Vectorization vs Broadcasting

These two concepts are related but different:

### Vectorization

Performing an operation on an entire array without explicitly writing a Python loop.

```python
prices * 2
```

### Broadcasting

Allows NumPy arrays with different shapes—or a scalar and an array—to work together.

```python
prices * discount
```

Here, `discount` is conceptually applied to every element.

### Simple Difference 💡

```text
Vectorization → Avoid Python loops

Broadcasting → Makes different-sized arrays compatible
               for operations
```

For your previous example:

```python
final_prices = prices - (prices * discount / 100)
```

**Vectorization** allows the calculation to run on the entire array, while **broadcasting** allows the scalar `discount` to operate with every element of `prices`.

# **Handaling Missing Value iN Numpy**

# Handling Missing Values in NumPy

In NumPy, missing values are commonly represented using:

```python
np.nan
```

`NaN` means **Not a Number**.

First, import NumPy:

```python
import numpy as np
```

---

## 1. Creating an Array with Missing Values

```python
arr = np.array([10, 20, np.nan, 40, 50])

print(arr)
```

Output:

```text
[10. 20. nan 40. 50.]
```

---

## 2. Finding Missing Values

Use `np.isnan()`:

```python
arr = np.array([10, 20, np.nan, 40, np.nan])

print(np.isnan(arr))
```

Output:

```text
[False False  True False  True]
```

`True` means that the value is missing.

### Count Missing Values

```python
np.isnan(arr).sum()
```

Output:

```text
2
```

---

## 3. Remove Missing Values

You can remove `NaN` values using:

```python
clean_arr = arr[~np.isnan(arr)]

print(clean_arr)
```

Output:

```text
[10. 20. 40.]
```

### How does this work?

```python
np.isnan(arr)
```

gives:

```text
[False False True False True]
```

The `~` operator reverses the values:

```text
[True True False True False]
```

Then NumPy selects only the `True` positions.

---

## 4. Replace Missing Values

### Replace with Zero

```python
arr = np.array([10, 20, np.nan, 40])

arr[np.isnan(arr)] = 0

print(arr)
```

Output:

```text
[10. 20.  0. 40.]
```

---

## 5. Replace Missing Values with the Mean

```python
arr = np.array([10, 20, np.nan, 40, 50])

mean_value = np.nanmean(arr)

arr[np.isnan(arr)] = mean_value

print(arr)
```

Output:

```text
[10. 20. 30. 40. 50.]
```

### Why `np.nanmean()`?

Normal mean:

```python
np.mean(arr)
```

returns:

```text
nan
```

because the array contains a missing value.

But:

```python
np.nanmean(arr)
```

ignores `NaN` values.

```text
(10 + 20 + 40 + 50) / 4 = 30
```

---

# Other Useful Functions

NumPy provides several functions that automatically ignore `NaN` values:

| Function         | Purpose              |
| ---------------- | -------------------- |
| `np.nanmean()`   | Mean ignoring NaN    |
| `np.nansum()`    | Sum ignoring NaN     |
| `np.nanmax()`    | Maximum ignoring NaN |
| `np.nanmin()`    | Minimum ignoring NaN |
| `np.nanmedian()` | Median ignoring NaN  |

### Example

```python
arr = np.array([10, 20, np.nan, 40])

print(np.nansum(arr))
```

Output:

```text
70.0
```

## Summary

```text
Find missing values    → np.isnan()
Count missing values   → np.isnan(arr).sum()
Remove missing values  → arr[~np.isnan(arr)]
Replace missing values → arr[np.isnan(arr)] = value
Ignore NaN in mean     → np.nanmean(arr)
```

💡 **Note:** For more advanced missing-data handling, especially in datasets with text, dates, and columns, **Pandas** is generally more convenient than NumPy.
