```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
titanic=pd.read_csv('train.csv')
```

```python
flights=sns.load_dataset('flights')
```

```python
sns.barplot(x=titanic['Pclass'], y=titanic['Age'],hue=titanic['Sex'])
```

**Box -Plot**

```python
sns.boxplot(x=titanic['Sex'], y=titanic['Age'])
```

**4.Dist-Plot**

```python
sns.displot(titanic[titanic['Survived']==0]['Age'])
sns.displot(titanic[titanic['Survived']==1]['Age'])
```

**5.Heatmap(Categorical -categorical)**

```python
  #in every pclass how much of dead or how much survived
pd.crosstab(titanic['Pclass'],titanic['Survived'])
```

```python
sns.heatmap(pd.crosstab(titanic['Pclass'],titanic['Survived']))
```

```python
titanic['Survived'] = pd.to_numeric(titanic['Survived'], errors='coerce')
titanic.groupby('Pclass')['Survived'].mean()*100
```

```python
(titanic.groupby('Pclass')['Survived'].mean()*100).plot(kind='bar')
```

**5.ClusterMap (categorical -categorical)**

```python
pd.crosstab(titanic['SibSp'],titanic['Survived'])
```

```python
sns.heatmap(pd.crosstab(titanic['Parch'],titanic['Survived']))
```
