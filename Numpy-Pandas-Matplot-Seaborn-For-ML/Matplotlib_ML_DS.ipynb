**Annotation**

In Matplotlib, annotation is used to add an explanatory label or note to a specific point on a graph.

It solves the problem of:

“How can I tell the viewer what an important point in my graph represents?”

```python


x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 40, 25]

plt.plot(x, y)

plt.annotate(
    "Highest",
    xy=(4, 40),
    xytext=(3, 45),
    arrowprops=dict(arrowstyle="->")
)

plt.show()
```
