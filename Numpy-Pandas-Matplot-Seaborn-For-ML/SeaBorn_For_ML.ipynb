**Relational plot**

```python
#To see the statical relation between 2 or more variable
#bivariate analysis

#1.plots under this
  #1.Scatterplot
  #2.lineplot
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
tips=sns.load_dataset('tips')
tips
```

```python
#scatter plot -> Axis level function
sns.scatterplot(data=tips,x='total_bill',y='tip',hue='sex',style='time',size='size')
```

```python
#figure level should ->Square shape
sns.relplot(data=tips,x='total_bill',y='tip',kind='scatter',hue='sex')
#by changing the kind like pie or line give diff graph
```

**Line Plot**

```python
gap=px.data.gapminder()
temp_df=gap[gap['country']=='China']
temp_df
```

```python
#axis plot
sns.lineplot(data=temp_df,x='year',y='lifeExp')
```

```python
#figure level
sns.relplot(data=temp_df,x='year',y='lifeExp',kind='line')
```

Anylasis of 3 countary data  

```python
temp_df=gap[gap['country'].isin(['India','Brazil','Germany'])]
temp_df
```

```python
sns.relplot(kind='line',data=temp_df,x='year',y='lifeExp',hue='country',style='continent')
```

```python
#facet plot
tips
```

```python
  #facet plot ->figure level function . work with relplot but not work with scatter plot
sns.relplot(data=tips,x='total_bill',y='tip',kind='scatter',hue='sex',col='sex',row='day')
```

**Distribution Plot**

A distribution plot is used to understand how numerical data is distributed—where values are concentrated, how spread out they are, and whether the data is skewed.

Application

(find distribution, univariate anylasis,range of observation,centeral tendency,is data is bimodel, are there outliers)
